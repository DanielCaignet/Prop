```python
def _linear_map(x, in_min, in_max, out_min, out_max, eps=1e-10):
    x_np = np.asarray(x)
    in_min_np = np.asarray(in_min)
    in_max_np = np.asarray(in_max)

    # Evitar división por cero si in_min == in_max
    denominator = in_max_np - in_min_np + eps
    if np.isscalar(denominator) and np.abs(denominator) < eps:
        denominator = eps # Prevent division by zero

    x_map = out_min + (x_np - in_min_np) * (out_max - out_min) / denominator
    return np.clip(x_map, out_min, out_max)

class HLogScaler:
    """ Scaler para Altura (Height) """
    def __init__(self, eps: float = 1e-10):
        self.eps = eps
        self.h_min = None
        self.h_max = None
        self.y_min = None
        self.y_max = None

    def fit(self, h: np.ndarray):
        h = np.asarray(h)
        self.h_min = h.min()
        self.h_max = h.max()

        if np.abs(self.h_min - self.h_max) < self.eps:
          self.h_max += self.eps

        hTN = _linear_map(h, self.h_min, self.h_max, 0.1, 1.0, self.eps)

        yLog = np.log10(hTN) # hTN ya está > 0

        self.y_min = yLog.min()
        self.y_max = yLog.max()

        if np.abs(self.y_min - self.y_max) < self.eps:
          self.y_max += self.eps

        return self

    def transform(self, h: np.ndarray) -> torch.Tensor:
        h_np = np.asarray(h)

        if self.h_min is None or self.h_max is None:
            raise RuntimeError("El scaler debe ser ajustado ('fit') antes de transformar datos.")

        hTN = _linear_map(h_np, self.h_min, self.h_max, 0.1, 1.0, self.eps)
        yLog = np.log10(hTN)

        yNorm = _linear_map(yLog, self.y_min, self.y_max, -1.0, 1.0, self.eps)

        return torch.tensor(yNorm, dtype=torch.float32)

    def inverse_transform(self, yNorm: np.ndarray or torch.Tensor) -> np.ndarray:
        if isinstance(yNorm, torch.Tensor):
            yNorm = yNorm.cpu().numpy()

        yLog = _linear_map(yNorm, -1.0, 1.0, self.y_min, self.y_max, self.eps)

        hTN = 10.0 ** yLog

        h = _linear_map(hTN, 0.1, 1.0, self.h_min, self.h_max, self.eps)

        return h

class LSineLogScaler:
    """ Scaler para Longitud (Length) """
    def __init__(self, eps: float = 1e-8):
        self.eps = eps
        self.z_min = None
        self.z_max = None

    def fit(self, l: np.ndarray):
        l = np.asarray(l)
        z = np.log10(l + self.eps) # Añadido eps para evitar log(0)

        self.z_min = z.min()
        self.z_max = z.max()

        if np.abs(self.z_min - self.z_max) < self.eps:
          self.z_max += self.eps

        return self

    def transform(self, lengths: np.ndarray) -> torch.Tensor:
        lengths_np = np.asarray(lengths)
        z = np.log10(lengths_np + self.eps)
        z_norm = _linear_map(z, self.z_min, self.z_max, -1.0, 1.0, self.eps)
        y = np.sin(np.pi / 2 * z_norm)
        return torch.tensor(y, dtype=torch.float32)

    def inverse_transform(self, y: np.ndarray or torch.Tensor) -> np.ndarray:
        if isinstance(y, torch.Tensor):
            y = y.cpu().numpy()

        # Clip input a arcsin para evitar errores de dominio
        y_clipped = np.clip(y, -1.0 + self.eps , 1.0 - self.eps)

        z_norm = (2 / np.pi) * np.arcsin(y_clipped)
        z = _linear_map(z_norm, -1.0, 1.0, self.z_min, self.z_max, self.eps)

        lengths = 10 ** z

        return lengths
```

```python
class SimpleWaveletDWT:
    def __init__(self, wavelet: str = "db4", level: int = 4, mode: str = "symmetric"):
        if pywt is None:
            raise ImportError("pywt no está disponible. Instala 'PyWavelets' para usar SimpleWaveletDWT.")
        if level < 1:
            raise ValueError("level debe ser >= 1.")

        self.wavelet_name = wavelet
        self.level = int(level)
        self.mode = mode

        # Inicializa objeto wavelet y valida
        try:
            self.w_ = pywt.Wavelet(self.wavelet_name)
        except Exception as e:
            raise ValueError(f"Wavelet '{self.wavelet_name}' no reconocida por pywt: {e}")

    def min_length_required(self) -> int:
        """
        Estima la longitud mínima de entrada para obtener al menos 1 coeficiente
        en el nivel 'self.level', asumiendo 'mode' compatible con la fórmula.
        """
        # Fórmula iterativa basada en longitud del filtro de descomposición (dec_len)
        L = self.w_.dec_len
        min_len = 1
        for _ in range(self.level):
            min_len = (min_len - 1) * 2 + L
        return int(min_len)

    def transform(self, x: np.ndarray) -> tuple[np.ndarray, list[np.ndarray]]:
        """
        Aplica DWT multinivel a 'x' (1D).

        Parámetros
        ----------
        x : array-like
            Señal 1D (convertida internamente a float64).

        Retorna
        -------
        cA : np.ndarray
            Coeficientes de aproximación del nivel más alto (nivel 'self.level').
        cD_list : list[np.ndarray]
            Lista de coeficientes de detalle del mayor al menor nivel: [cD_L, ..., cD_1].
        """
        x_np = np.asarray(x, dtype=float).ravel()
        if x_np.size == 0:
            return np.array([], dtype=float), []

        # Verificación rápida de longitud
        if x_np.size < self.min_length_required():
            # Señal demasiado corta para el nivel pedido
            return np.array([], dtype=float), []

        try:
            coeffs = pywt.wavedec(x_np, self.w_, level=self.level, mode=self.mode)
            # coeffs = [cA_L, cD_L, cD_{L-1}, ..., cD_1]
            cA = coeffs[0]
            cD_list = coeffs[1:]  # ya viene ordenado de L -> 1
            return cA, cD_list
        except Exception:
            # Si pywt decide que el nivel es excesivo para la longitud real (dependiendo del modo),
            # se retorna vacío para que el caller decida qué hacer.
            return np.array([], dtype=float), []

    def transform_many(self, signals: list[np.ndarray]) -> list[tuple[np.ndarray, list[np.ndarray]]]:
        """
        Aplica 'transform' a una lista de señales, preservando el orden.

        Retorna
        -------
        results : list de tuplas
            Cada elemento es (cA, cD_list) como en 'transform'.
        """
        results = []
        for x in signals:
            results.append(self.transform(x))
        return results

class SimpleNormalizer:
    """
    Normaliza señales numéricas usando los valores mínimo y máximo
    obtenidos del conjunto de entrenamiento.

    Ejemplo de uso:
        scaler = SimpleNormalizer()
        scaler.fit(train_signals)
        x_norm = scaler.transform(x, target_range=(-1, 1))
    """
    def __init__(self, eps: float = 1e-8):
        self.eps = eps
        self.x_min = None
        self.x_max = None

    def fit(self, signals: list[np.ndarray]):
        """
        Ajusta el normalizador a partir de una lista de señales (entrenamiento).
        Calcula min y max globales.
        """
        if not signals:
            raise ValueError("La lista de señales está vacía.")
        z = np.concatenate([np.asarray(s, dtype=float).ravel() for s in signals])
        self.x_min = np.min(z)
        self.x_max = np.max(z)
        if np.abs(self.x_max - self.x_min) < self.eps:
            self.x_max += self.eps  # evita división por cero
        return self

    def transform(self, x: np.ndarray, target_range=(-1.0, 1.0)) -> np.ndarray:
        """
        Normaliza una señal con base en los valores min/max aprendidos.
        """
        if self.x_min is None or self.x_max is None:
            raise RuntimeError("Debe llamar a 'fit()' antes de 'transform()'.")
        return _linear_map(
            x,
            in_min=self.x_min,
            in_max=self.x_max,
            out_min=target_range[0],
            out_max=target_range[1],
            eps=self.eps
        )

    def inverse_transform(self, x_norm: np.ndarray, target_range=(-1.0, 1.0)) -> np.ndarray:
        """
        Desnormaliza una señal previamente escalada al rango target_range.
        """
        if self.x_min is None or self.x_max is None:
            raise RuntimeError("Debe llamar a 'fit()' antes de 'inverse_transform()'.")
        return _linear_map(
            x_norm,
            in_min=target_range[0],
            in_max=target_range[1],
            out_min=self.x_min,
            out_max=self.x_max,
            eps=self.eps
        )

ArrayLike = Union[np.ndarray, List[float]]

class TimeSeriesWindowizerPadLast:
    """
    Genera ventanas solapadas de tamaño W a partir de series 1D, rellenando la última ventana (cuando no alcanza W) con el último valor de la serie.

      Cada ventana está asociada con:
        - el índice de la serie original
        - el índice de la ventana dentro de la serie
        - la porción (start, end) dentro de la serie original
        - tiempos (opcional) si se entrega vector de tiempos
        - etiqueta(es) de la serie original (copiadas para cada ventana)

      Parámetros
      ----------
          window_size : int
              Tamaño de cada ventana (W).
          overlap : float
              Solapamiento en [0,1). Ejemplo: 0.5 => hop = W*(1-0.5).
          short_policy : {'single'}
              Para series más cortas que W: siempre se genera **una** ventana,
              rellenada con el último valor.
          pad_value_mode : {'last_value'}
              Modo de padding: en este caso sólo 'last_value' (relleno con último valor de la serie).
    """

    def __init__(self,
                 window_size: int,
                 overlap: float = 0.5):
        if not (0 <= overlap < 1):
            raise ValueError("overlap debe estar en [0,1).")
        if window_size < 1:
            raise ValueError("window_size debe ser >= 1.")

        self.W = int(window_size)
        self.overlap = float(overlap)
        self.hop = max(1, int(round(self.W * (1.0 - self.overlap))))

    def _starts_for_length(self, L: int) -> List[int]:
        """
        Genera los índices de inicio para ventanas dada longitud L,
        cubriendo hasta el final mediante padding si es necesario.
        """
        if L <= 0:
            return []
        if L < self.W:
            # Serie muy corta; generamos solo inicio 0
            return [0]

        starts = list(range(0, L - self.W + 1, self.hop))
        last_start = starts[-1] if starts else 0
        if last_start + self.W < L:
            # queda fragmento al final que no llena W: añadimos ventana final alineada al final-W
            starts.append(L - self.W)
        return starts

    def _get_window(self, x: np.ndarray, start: int) -> np.ndarray:
        """
        Extrae ventana x[start: start+W]. Si la señal no tiene W muestras
        (inicio demasiado cerca del final) o si la señal es más corta que W,
        se padéa con el **último valor de la serie**.
        """
        L = x.size
        end = start + self.W
        if start < 0:
            raise ValueError("start debe ser >= 0")
        if start >= L:
            # inicio fuera de la señal → ventana de W repetida con último valor
            last = x[-1] if L > 0 else 0.0
            return np.full((self.W,), last, dtype=float)

        if end <= L:
            return x[start:end].astype(float)
        # fragmento hasta L, luego padding
        slice_part = x[start:L].astype(float)
        last_val = x[-1]
        n_pad = end - L
        pad_part = np.full((n_pad,), last_val, dtype=float)
        return np.concatenate([slice_part, pad_part])

    def transform(self,
                  signals: List[ArrayLike],
                  labels: Optional[Union[List[float], np.ndarray, Dict[str, ArrayLike]]] = None,
                  times: Optional[List[ArrayLike]] = None
                  ) -> Tuple[np.ndarray, Dict[str, np.ndarray], Dict[str, np.ndarray]]:
        """
        Genera ventanas y devuelve (X, y, meta).

        signals : lista de arrays 1D (longitudes variables).
        labels  : (opcional) dict de etiquetas por serie, o 1D de escalares por serie.
        times   : (opcional) lista de arrays 1D de tiempos, uno por serie.

        Retornos
        --------
        X : np.ndarray de shape (N_windows, window_size)
        y : Dict[str, np.ndarray]  (cada clave con shape (N_windows,))
        meta : Dict[str, np.ndarray]
        """
        X_list = []
        y_dict: Dict[str, List[float]] = {}
        meta: Dict[str, List[Union[int, float]]] = {
            "sample_index": [],
            "window_index_within_sample": [],
            "start": [],
            "end": [],
            "orig_len": []
        }
        use_times = times is not None
        if use_times:
            meta["t_start"] = []
            meta["t_end"] = []

        # Preparar etiquetas
        if labels is None:
            labels_mode = "none"
            series_labels = None
        elif isinstance(labels, dict):
            labels_mode = "dict"
            for k, arr in labels.items():
                if len(arr) != len(signals):
                    raise ValueError(f"labels['{k}'] debe tener la misma longitud que signals.")
                y_dict[k] = []
            series_labels = labels
        else:
            labels_mode = "vector"
            labels_arr = np.asarray(labels, dtype=float)
            if labels_arr.shape[0] != len(signals):
                raise ValueError("labels debe tener la misma longitud que signals.")
            y_dict["label"] = []
            series_labels = labels_arr

        for i, x in enumerate(signals):
            x_np = np.asarray(x, dtype=float).ravel()
            L = x_np.size
            starts = self._starts_for_length(L)

            t = None
            if use_times:
                t = np.asarray(times[i], dtype=float).ravel()
                if t.size != L:
                    raise ValueError(f"times[{i}] debe tener la misma longitud que signals[{i}].")

            if labels_mode == "dict":
                series_i_labels = {k: float(series_labels[k][i]) for k in series_labels}
            elif labels_mode == "vector":
                series_i_label = float(series_labels[i])
            else:
                series_i_labels = None

            for wj, s in enumerate(starts):
                win = self._get_window(x_np, s)
                X_list.append(win)

                meta["sample_index"].append(i)
                meta["window_index_within_sample"].append(wj)
                meta["start"].append(s)
                meta["end"].append(s + self.W)
                meta["orig_len"].append(L)

                if use_times:
                    # definimos t_start como t[s], t_end como t[min(s+W-1, L-1)]
                    s_eff = min(s, L-1)
                    e_eff = min(s + self.W - 1, L-1)
                    meta["t_start"].append(t[s_eff])
                    meta["t_end"].append(t[e_eff])

                if labels_mode == "dict":
                    for k, v in series_i_labels.items():
                        y_dict[k].append(v)
                elif labels_mode == "vector":
                    y_dict["label"].append(series_i_label)

        if not X_list:
            X = np.empty((0, self.W), dtype=float)
            y_out = {k: np.empty((0,), dtype=float) for k in y_dict}
            meta_out = {k: np.empty((0,), dtype=float) for k in meta}
            return X, y_out, meta_out

        X = np.vstack(X_list).astype(float)
        y_out = {k: np.asarray(v, dtype=float) for k, v in y_dict.items()}
        meta_out = {k: np.asarray(v) for k, v in meta.items()}

        return X, y_out, meta_out
```

```python
"""#clase v1
class VehicleLHDataset(Dataset):

    Dataset que produce ventanas solapadas de cA (coef. de aproximación) ya normalizados.
      Pipeline por muestra:
          raw_power -> DWT (cA) -> normalización (fit en train) -> ventanas con pad-last
      Cada ítem corresponde a UNA ventana y trae sus labels (length, height) y metadatos.

      Requisitos (inyectar instancias ya creadas/ajustadas):
        - dwt: objeto con .transform(x) -> (cA, cD_list)
        - normalizer: objeto con .transform(x, target_range) -> np.ndarray
        - windowizer: objeto con .transform(signals, labels, times) -> (X, y, meta)
                      Debe padear la última ventana con el último valor (como el TimeSeriesWindowizerPadLast que definimos)
        - length_scaler: LSineLogScaler (fit con train)
        - hight_scaler : HLogScaler (fit con train)

      Retorno de __getitem__(i):
        - x_win: torch.FloatTensor (W,)    ventana normalizada
        - y_len: torch.FloatTensor ()      length escalado
        - y_hig: torch.FloatTensor ()      height escalado
        - meta : dict con claves: sample_index, window_index_within_sample, start, end, orig_len
                (y t_start/t_end si tu windowizer las generó)


    def __init__(self,
                 indices,
                 raw_powers,
                 raw_lengths,
                 raw_hights,
                 dwt,
                 normalizer,
                 windowizer,
                 length_scaler,
                 hight_scaler,
                 target_range=(-1.0, 1.0)):
        super().__init__()
        self.target_range = target_range
        self.length_scaler = length_scaler
        self.hight_scaler = hight_scaler

        # --- 1) Seleccionar subconjunto por indices ---
        self.idx_map = np.asarray(indices, dtype=int)
        powers_subset = [np.asarray(raw_powers[i], dtype=float).ravel() for i in self.idx_map]
        lengths_sub   = np.asarray([raw_lengths[i] for i in self.idx_map], dtype=float)
        hights_sub    = np.asarray([raw_hights[i]  for i in self.idx_map], dtype=float)

        # --- 2) DWT -> cA (con fallback si vacío) ---
        ca_list = []
        for x in powers_subset:
            cA, _ = dwt.transform(x)
            if cA is None or np.size(cA) == 0:
                # Fallback mínimo: usar el último valor de la señal (o 0 si señal vacía)
                last_val = float(x[-1]) if x.size > 0 else 0.0
                cA = np.array([last_val], dtype=float)
            else:
                cA = np.asarray(cA, dtype=float).ravel()
            ca_list.append(cA)

        # --- 3) Normalización de cA con scaler entrenado en TRAIN ---
        ca_norm_list = [normalizer.transform(ca, target_range=self.target_range) for ca in ca_list]

        # --- 4) Enventanado + etiquetas por serie (copiadas por ventana) ---
        labels_dict = {"length": lengths_sub, "height": hights_sub}
        X, y, meta = windowizer.transform(signals=ca_norm_list,
                                          labels=labels_dict,
                                          times=None)  # opcionalmente podrías derivar tiempos para cA si así lo defines

        # Guardar en el dataset
        self.X = torch.tensor(X, dtype=torch.float32)                    # (Nw, W)
        self.y_len_raw = y["length"].astype(float)                       # (Nw,)
        self.y_hig_raw = y["height"].astype(float)                       # (Nw,)
        self.meta = meta                                                 # dict de arrays numpy

        # También conservamos un mapeo por ventana → índice global de evento
        # meta['sample_index'] ya es el índice dentro del subset; lo mapeamos al índice global original:
        self.meta_global_sample_index = self.idx_map[self.meta["sample_index"]]

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, i):
        # Ventana normalizada (cA)
        x_win = self.X[i]  # torch.FloatTensor (W,)

        # Etiquetas escaladas (aplicar en tiempo de acceso; son escalares)
        l_val = np.array([self.y_len_raw[i]], dtype=float)
        h_val = np.array([self.y_hig_raw[i]], dtype=float)

        l_tensor = self.length_scaler.transform(l_val).reshape(-1)  # torch tensor
        h_tensor = self.hight_scaler.transform(h_val).reshape(-1)

        # Metadatos
        meta = {
            "global_sample_index": int(self.meta_global_sample_index[i]),
            "sample_index":       int(self.meta["sample_index"][i]),
            "window_index":       int(self.meta["window_index_within_sample"][i]),
            "start":              int(self.meta["start"][i]),
            "end":                int(self.meta["end"][i]),
            "orig_len":           int(self.meta["orig_len"][i]),
        }
        # Añadir tiempos si existen
        if "t_start" in self.meta and "t_end" in self.meta:
            meta["t_start"] = float(self.meta["t_start"][i])
            meta["t_end"]   = float(self.meta["t_end"][i])

        # Devolver tensores escalares (0-D o (1,)), por simplicidad los dejamos como (1,)
        return x_win, l_tensor.squeeze(0), h_tensor.squeeze(0), meta
"""
#clase v2
class VehicleLHDataset(Dataset):
    """
      Dataset que produce ventanas solapadas de cA normalizados y añade como
      característica la longitud original de la serie (normalizada).

      Retorno de __getitem__(i):
        - x_win  : torch.FloatTensor (W,)      ventana normalizada (cA)
        - len_ft : torch.FloatTensor ()        longitud original normalizada del evento
        - y_len  : torch.FloatTensor ()        length (etiqueta) escalada
        - y_hig  : torch.FloatTensor ()        height (etiqueta) escalada
        - meta   : dict                        metadatos por ventana
    """

    def __init__(self,
                 indices,
                 raw_powers,
                 raw_lengths,
                 raw_hights,
                 dwt,
                 normalizer,          # para cA (fit en train)
                 windowizer,          # pad-last con último valor
                 length_scaler,       # LSineLogScaler (fit en train)
                 hight_scaler,        # HLogScaler (fit en train)
                 len_feat_normalizer, # SimpleNormalizer (fit SOLO con longitudes de serie del split de TRAIN)
                 target_range=(-1.0, 1.0)):
        super().__init__()
        self.target_range = target_range
        self.length_scaler = length_scaler
        self.hight_scaler  = hight_scaler
        self.len_feat_norm = len_feat_normalizer

        # 1) Subconjunto por índices
        self.idx_map = np.asarray(indices, dtype=int)
        powers_subset = [np.asarray(raw_powers[i], dtype=float).ravel() for i in self.idx_map]
        lengths_sub   = np.asarray([raw_lengths[i] for i in self.idx_map], dtype=float)
        hights_sub    = np.asarray([raw_hights[i]  for i in self.idx_map], dtype=float)

        # 2) DWT -> cA (fallback si vacío)
        ca_list = []
        for x in powers_subset:
            cA, _ = dwt.transform(x)
            if cA is None or np.size(cA) == 0:
                last_val = float(x[-1]) if x.size > 0 else 0.0
                cA = np.array([last_val], dtype=float)
            else:
                cA = np.asarray(cA, dtype=float).ravel()
            ca_list.append(cA)

        # 3) Normalizar cA con normalizador ya ajustado
        ca_norm_list = [normalizer.transform(ca, target_range=self.target_range) for ca in ca_list]

        # 4) Enventanar + etiquetas por serie
        labels_dict = {"length": lengths_sub, "height": hights_sub}
        X, y, meta = windowizer.transform(signals=ca_norm_list, labels=labels_dict, times=None)

        # 5) Guardar tensores y etiquetas crudas (por ventana)
        self.X = torch.tensor(X, dtype=torch.float32)              # (Nw, W)
        self.y_len_raw = y["length"].astype(float)                 # (Nw,)
        self.y_hig_raw = y["height"].astype(float)                 # (Nw,)
        self.meta = meta                                           # dict con 'orig_len' por ventana

        # 6) Característica nueva: longitud original de la serie (por ventana) NORMALIZADA
        #    meta['orig_len'] está por ventana; lo transformamos con len_feat_normalizer
        orig_len_vec = self.meta["orig_len"].astype(float)         # (Nw,)
        len_feat_norm = self.len_feat_norm.transform(orig_len_vec) # np.ndarray (Nw,)
        self.len_feat = torch.tensor(len_feat_norm, dtype=torch.float32)

        # 7) Mapeo a índice global de evento (útil en metadatos)
        self.meta_global_sample_index = self.idx_map[self.meta["sample_index"]]

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, i):
        x_win = self.X[i]                         # (W,)
        len_ft = self.len_feat[i]                 # () escalar normalizado

        # Etiquetas (escaladas al vuelo)
        l_tensor = self.length_scaler.transform(np.array([self.y_len_raw[i]], dtype=float)).reshape(-1)
        h_tensor = self.hight_scaler.transform(np.array([self.y_hig_raw[i]], dtype=float)).reshape(-1)

        meta = {
            "global_sample_index": int(self.meta_global_sample_index[i]),
            "sample_index":       int(self.meta["sample_index"][i]),
            "window_index":       int(self.meta["window_index_within_sample"][i]),
            "start":              int(self.meta["start"][i]),
            "end":                int(self.meta["end"][i]),
            "orig_len":           int(self.meta["orig_len"][i]),
        }
        if "t_start" in self.meta and "t_end" in self.meta:
            meta["t_start"] = float(self.meta["t_start"][i])
            meta["t_end"]   = float(self.meta["t_end"][i])

        # NUEVO ORDEN: x_win, len_feat, y_len, y_hig, meta
        return x_win, len_ft, l_tensor.squeeze(0), h_tensor.squeeze(0), meta
```

```python
# ===========================================================
# Preparación de features compartidos para todos los modelos
# ===========================================================
from dataclasses import dataclass

import pandas as pd
from scipy import signal, stats
from sklearn.preprocessing import StandardScaler

EPS = 1e-12

@dataclass
class PreparedWindowData:
    X_train: pd.DataFrame
    X_test: pd.DataFrame
    y_train_scaled: np.ndarray
    y_test_scaled: np.ndarray
    y_test_real: np.ndarray
    groups_train: np.ndarray
    groups_test: np.ndarray

def dataset_to_arrays_with_groups(ds, length_scaler, hight_scaler):
    Xw, Xm, Ysc, Yreal, groups = [], [], [], [], []
    for i in range(len(ds)):
        x_win, len_ft, y_len_sc, y_hig_sc, meta = ds[i]
        start, end = meta["start"], meta["end"]
        widx, olen = meta["window_index"], meta["orig_len"]
        gidx = meta.get("global_sample_index", meta["sample_index"])

        Xw.append(x_win.numpy())
        Xm.append([float(len_ft.item()), float(start), float(end), float(widx), float(olen)])
        Ysc.append([float(y_len_sc.item()), float(y_hig_sc.item())])

        y_len_real = length_scaler.inverse_transform(np.array([y_len_sc.item()]))[0]
        y_hig_real = hight_scaler.inverse_transform(np.array([y_hig_sc.item()]))[0]
        Yreal.append([float(y_len_real), float(y_hig_real)])
        groups.append(int(gidx))

    return (
        np.asarray(Xw, dtype=np.float32),
        np.asarray(Xm, dtype=np.float32),
        np.asarray(Ysc, dtype=np.float32),
        np.asarray(Yreal, dtype=np.float32),
        np.asarray(groups, dtype=int),
    )

def _compute_window_ratios(Xm: np.ndarray) -> np.ndarray:
    denom = Xm[:, 4] + 1e-6
    r_start = Xm[:, 1] / denom
    r_end = Xm[:, 2] / denom
    r_center = 0.5 * (Xm[:, 1] + Xm[:, 2]) / denom
    return np.column_stack([r_start, r_end, r_center]).astype(np.float32)

def _run_lengths(mask: np.ndarray) -> tuple[float, float]:
    if mask.sum() == 0:
        return 0.0, 0.0
    dif = np.diff(mask.astype(np.int32), prepend=0, append=0)
    starts = np.where(dif == 1)[0]
    ends = np.where(dif == -1)[0]
    lens = (ends - starts).astype(np.float64)
    return float(np.mean(lens)), float(len(lens))

def _time_features_power(x: np.ndarray, M: int, max_len: int):
    x = np.asarray(x, dtype=np.float64).ravel()
    if x.size == 0:
        x = np.zeros(1, dtype=np.float64)
        M = 1
    dx = np.diff(x, prepend=x[0])
    rms = float(np.sqrt(np.mean(x**2)))
    peak = float(np.max(np.abs(x)))
    crest = float(peak / (rms + EPS))
    loadf = float(np.mean(x) / (peak + EPS))
    p50, p75, p90 = np.percentile(x, [50, 75, 90])
    duty50 = float(np.mean(x > p50))
    duty75 = float(np.mean(x > p75))
    duty90 = float(np.mean(x > p90))
    mean_on, n_on = _run_lengths(x > p75)
    xc = x - x.mean()
    ac = signal.correlate(xc, xc, mode="full")
    ac = ac[ac.size // 2 :]
    ac = ac / (ac[0] + EPS)
    if ac.size > 2:
        dom_lag = int(np.argmax(ac[1 : min(len(ac), len(x) // 2 + 1)]) + 1)
        dom_lag_n = dom_lag / len(x)
        ac_energy = float(np.sum(ac**2))
    else:
        dom_lag_n, ac_energy = 0.0, 0.0
    idx = np.linspace(0, 1, len(x))
    slope = float(np.polyfit(idx, x, 1)[0]) if len(x) > 1 else 0.0
    return {
        "dur_norm": float(M / (max_len + EPS)),
        "mean": float(np.mean(x)),
        "std": float(np.std(x)),
        "min": float(np.min(x)),
        "max": float(np.max(x)),
        "ptp": float(np.ptp(x)),
        "median": float(np.median(x)),
        "q05": float(np.quantile(x, 0.05)),
        "q95": float(np.quantile(x, 0.95)),
        "iqr": float(np.quantile(x, 0.75) - np.quantile(x, 0.25)),
        "rms": rms,
        "energy": float(np.sum(x**2)),
        "skew": float(stats.skew(x, bias=False)) if len(x) > 2 else 0.0,
        "kurt": float(stats.kurtosis(x, bias=False)) if len(x) > 3 else 0.0,
        "zcr": float(np.mean(np.abs(np.diff(np.signbit(x - np.mean(x)))))),
        "max_abs_dx": float(np.max(np.abs(dx))),
        "mad_dx": float(np.mean(np.abs(dx))),
        "crest_factor": crest,
        "load_factor": loadf,
        "duty50": duty50,
        "duty75": duty75,
        "duty90": duty90,
        "mean_on": mean_on,
        "n_on": n_on,
        "ac_dom_lag_n": float(dom_lag_n),
        "ac_energy": ac_energy,
        "trend_slope": slope,
    }

def _spectral_features_power(x: np.ndarray, nperseg: int = 256, noverlap: int = 128):
    x = np.asarray(x, dtype=np.float64).ravel()
    if x.size == 0:
        x = np.zeros(1, dtype=np.float64)
    x = x - np.mean(x)
    nperseg = min(nperseg, len(x))
    noverlap = min(noverlap, max(0, len(x) // 2))
    if nperseg < 2:
        nperseg = len(x)
    f, pxx = signal.welch(x, fs=1.0, nperseg=nperseg, noverlap=noverlap)
    pxx = np.maximum(pxx, EPS)
    P = float(np.sum(pxx))
    centroid = float(np.sum(f * pxx) / (P + EPS))
    bandwidth = float(np.sqrt(np.sum(((f - centroid) ** 2) * pxx) / (P + EPS)))
    dom_idx = int(np.argmax(pxx))
    dom_f = float(f[dom_idx])
    dom_pow = float(pxx[dom_idx])
    cs = np.cumsum(pxx) / (P + EPS)
    roll85 = float(f[int(np.searchsorted(cs, 0.85, side="left"))])
    flatness = float(np.exp(np.mean(np.log(pxx))) / (np.mean(pxx) + EPS))
    mask = (f > 0) & np.isfinite(f)
    if mask.sum() > 2:
        slope = float(np.polyfit(np.log(f[mask]), np.log(pxx[mask]), 1)[0])
    else:
        slope = 0.0
    edges = [0.00, 0.05, 0.15, 0.30, 0.50]
    bands = [
        float(np.sum(pxx[(f >= a) & (f < b)]) / (P + EPS))
        for a, b in zip(edges[:-1], edges[1:])
    ]
    while len(bands) < 4:
        bands.append(0.0)
    return {
        "spec_power": P,
        "spec_centroid": centroid,
        "spec_bandwidth": bandwidth,
        "spec_dom_f": dom_f,
        "spec_dom_pow": dom_pow,
        "spec_rolloff85": roll85,
        "spec_flatness": flatness,
        "spec_loglog_slope": slope,
        "spec_band1": bands[0],
        "spec_band2": bands[1],
        "spec_band3": bands[2],
        "spec_band4": bands[3],
        "spec_low_high_ratio": (bands[0] + EPS) / (bands[3] + EPS),
    }

def _build_event_features_df(indices: np.ndarray, raw_powers, max_len_ref: int) -> pd.DataFrame:
    rows = []
    for ev_id in indices:
        x = np.asarray(raw_powers[int(ev_id)], dtype=np.float64).ravel()
        M = len(x)
        tf = _time_features_power(x, M=M, max_len=max_len_ref)
        sf = _spectral_features_power(x)
        feats = {**tf, **sf, "event_id": int(ev_id), "M_abs": float(M)}
        rows.append(feats)
    df = pd.DataFrame(rows).set_index("event_id").sort_index()
    return df

def _assemble_feature_frame(Xw: np.ndarray, Xm_scaled: np.ndarray, R_scaled: np.ndarray) -> pd.DataFrame:
    W = Xw.shape[1]
    win_cols = [f"w{i}" for i in range(W)]
    meta_cols = ["len_ft_norm", "start_s", "end_s", "window_idx", "orig_len_std"]
    ratio_cols = ["r_start", "r_end", "r_center"]
    data = np.hstack([Xw, Xm_scaled, R_scaled]).astype(np.float32)
    return pd.DataFrame(data, columns=win_cols + meta_cols + ratio_cols)

def prepare_window_data(train_ds, test_ds, length_scaler, hight_scaler, raw_powers, train_indices, test_indices):
    Xw_tr, Xm_tr, Ysc_tr, Yreal_tr, groups_tr = dataset_to_arrays_with_groups(train_ds, length_scaler, hight_scaler)
    Xw_te, Xm_te, Ysc_te, Yreal_te, groups_te = dataset_to_arrays_with_groups(test_ds, length_scaler, hight_scaler)

    meta_scaler = StandardScaler().fit(Xm_tr[:, 1:])
    Xm_tr_scaled = np.hstack([Xm_tr[:, [0]], meta_scaler.transform(Xm_tr[:, 1:])])
    Xm_te_scaled = np.hstack([Xm_te[:, [0]], meta_scaler.transform(Xm_te[:, 1:])])

    R_tr = _compute_window_ratios(Xm_tr)
    R_te = _compute_window_ratios(Xm_te)
    ratio_scaler = StandardScaler().fit(R_tr)
    R_tr_s = ratio_scaler.transform(R_tr)
    R_te_s = ratio_scaler.transform(R_te)

    df_tr_base = _assemble_feature_frame(Xw_tr, Xm_tr_scaled, R_tr_s)
    df_te_base = _assemble_feature_frame(Xw_te, Xm_te_scaled, R_te_s)

    max_len_train = max(len(raw_powers[int(i)]) for i in train_indices)
    df_ev_tr = _build_event_features_df(train_indices, raw_powers, max_len_train)
    df_ev_te = _build_event_features_df(test_indices, raw_powers, max_len_train)

    Fev_tr = df_ev_tr.loc[groups_tr]
    Fev_te = df_ev_te.loc[groups_te]

    ev_scaler = StandardScaler().fit(Fev_tr)
    Fev_tr_s = ev_scaler.transform(Fev_tr)
    Fev_te_s = ev_scaler.transform(Fev_te)

    ev_cols = [f"ev_{c}" for c in df_ev_tr.columns]
    df_ev_tr_windows = pd.DataFrame(Fev_tr_s, columns=ev_cols)
    df_ev_te_windows = pd.DataFrame(Fev_te_s, columns=ev_cols)

    X_train = pd.concat([df_tr_base.reset_index(drop=True), df_ev_tr_windows.reset_index(drop=True)], axis=1)
    X_test = pd.concat([df_te_base.reset_index(drop=True), df_ev_te_windows.reset_index(drop=True)], axis=1)

    return PreparedWindowData(
        X_train=X_train,
        X_test=X_test,
        y_train_scaled=Ysc_tr,
        y_test_scaled=Ysc_te,
        y_test_real=Yreal_te,
        groups_train=groups_tr,
        groups_test=groups_te,
    )

def evaluate_real_units(y_real_true, y_len_pred_sc, y_hig_pred_sc):
    y_len_pred = length_scaler.inverse_transform(np.asarray(y_len_pred_sc))
    y_hig_pred = hight_scaler.inverse_transform(np.asarray(y_hig_pred_sc))
    y_pred_real = np.stack([y_len_pred, y_hig_pred], axis=1)
    mae = np.mean(np.abs(y_real_true - y_pred_real), axis=0)
    mse = np.mean((y_real_true - y_pred_real) ** 2, axis=0)
    r2_len = r2_score(y_real_true[:, 0], y_pred_real[:, 0])
    r2_hig = r2_score(y_real_true[:, 1], y_pred_real[:, 1])
    return {
        "MAE_len": mae[0],
        "MAE_hig": mae[1],
        "MAE_avg": float(mae.mean()),
        "MSE_len": mse[0],
        "MSE_hig": mse[1],
        "MSE_avg": float(mse.mean()),
        "R2_len": r2_len,
        "R2_hig": r2_hig,
        "R2_avg": float((r2_len + r2_hig) / 2.0),
    }

prepared_data = prepare_window_data(train_ds, test_ds, length_scaler, hight_scaler, raw_powers, trainval_idx, test_idx)
print("Matrices de features (train/test):", prepared_data.X_train.shape, prepared_data.X_test.shape)
```

## LightGBM

```python
# Entrenamiento LightGBM con GroupKFold usando el mismo set de features
from lightgbm import LGBMRegressor, early_stopping, log_evaluation
from sklearn.model_selection import GroupKFold

def train_lightgbm_multioutput(
    prepared: PreparedWindowData,
    params: Optional[Dict] = None,
    n_splits: int = 5,
    es_rounds: int = 300,
    seed: int = SEED,
):
    base_params = dict(
        objective="regression",
        learning_rate=0.03,
        n_estimators=10000,
        num_leaves=127,
        max_depth=-1,
        subsample=0.85,
        colsample_bytree=0.8,
        min_child_samples=20,
        reg_lambda=1.0,
        random_state=seed,
    )
    if params:
        base_params.update(params)

    folds = list(GroupKFold(n_splits=n_splits).split(prepared.X_train, groups=prepared.groups_train))

    models_len, models_hig = [], []
    for tr_idx, va_idx in folds:
        model_len = LGBMRegressor(**base_params)
        model_len.fit(
            prepared.X_train.iloc[tr_idx],
            prepared.y_train_scaled[tr_idx, 0],
            eval_set=[(prepared.X_train.iloc[va_idx], prepared.y_train_scaled[va_idx, 0])],
            eval_metric="l2",
            callbacks=[
                early_stopping(es_rounds, first_metric_only=True, verbose=False),
                log_evaluation(period=0),
            ],
        )
        models_len.append(model_len)

        model_hig = LGBMRegressor(**base_params)
        model_hig.fit(
            prepared.X_train.iloc[tr_idx],
            prepared.y_train_scaled[tr_idx, 1],
            eval_set=[(prepared.X_train.iloc[va_idx], prepared.y_train_scaled[va_idx, 1])],
            eval_metric="l2",
            callbacks=[
                early_stopping(es_rounds, first_metric_only=True, verbose=False),
                log_evaluation(period=0),
            ],
        )
        models_hig.append(model_hig)

    pred_len = np.mean([m.predict(prepared.X_test) for m in models_len], axis=0)
    pred_hig = np.mean([m.predict(prepared.X_test) for m in models_hig], axis=0)
    return {
        "models_len": models_len,
        "models_hig": models_hig,
        "pred_len_scaled": pred_len,
        "pred_hig_scaled": pred_hig,
    }

lgbm_results = train_lightgbm_multioutput(prepared_data)
metrics_lgbm_windows = evaluate_real_units(
    prepared_data.y_test_real,
    lgbm_results["pred_len_scaled"],
    lgbm_results["pred_hig_scaled"],
)
print("LightGBM (métricas por ventana en unidades reales):", metrics_lgbm_windows)
```

## XGBoost

```python
# Entrenamiento XGBoost con el mismo set de features y GroupKFold
import xgboost as xgb
from sklearn.model_selection import GroupKFold

def _train_xgb_target(
    X_train: np.ndarray,
    y: np.ndarray,
    folds,
    params: Dict,
    num_boost_round: int,
    es_rounds: int,
):
    boosters = []
    for tr_idx, va_idx in folds:
        dtrain = xgb.DMatrix(X_train[tr_idx], label=y[tr_idx])
        dvalid = xgb.DMatrix(X_train[va_idx], label=y[va_idx])
        booster = xgb.train(
            params=params,
            dtrain=dtrain,
            num_boost_round=num_boost_round,
            evals=[(dvalid, "valid")],
            early_stopping_rounds=es_rounds,
            verbose_eval=False,
        )
        boosters.append(booster)
    return boosters

def _predict_xgb_ensemble(boosters, X_test: np.ndarray) -> np.ndarray:
    dtest = xgb.DMatrix(X_test)
    preds = []
    for booster in boosters:
        best_iteration = getattr(booster, "best_iteration", None)
        if best_iteration is not None:
            preds.append(booster.predict(dtest, iteration_range=(0, best_iteration + 1)))
        else:
            ntree = getattr(booster, "best_ntree_limit", 0)
            if ntree:
                preds.append(booster.predict(dtest, ntree_limit=ntree))
            else:
                preds.append(booster.predict(dtest))
    return np.mean(np.column_stack(preds), axis=0)

def train_xgboost_multioutput(
    prepared: PreparedWindowData,
    params_len: Optional[Dict] = None,
    params_hig: Optional[Dict] = None,
    n_splits: int = 5,
    num_boost_round_len: int = 10000,
    num_boost_round_hig: int = 8000,
    es_rounds_len: int = 400,
    es_rounds_hig: int = 300,
    seed: int = SEED,
):
    base_len = {
        "objective": "reg:squarederror",
        "eval_metric": "rmse",
        "eta": 0.025,
        "max_depth": 9,
        "min_child_weight": 3,
        "subsample": 0.85,
        "colsample_bytree": 0.9,
        "lambda": 1.5,
        "alpha": 0.0,
        "tree_method": "hist",
        "seed": seed,
    }
    base_hig = {
        "objective": "reg:squarederror",
        "eval_metric": "rmse",
        "eta": 0.05,
        "max_depth": 6,
        "subsample": 0.9,
        "colsample_bytree": 0.8,
        "lambda": 1.0,
        "alpha": 0.0,
        "tree_method": "hist",
        "seed": seed,
    }
    if params_len:
        base_len.update(params_len)
    if params_hig:
        base_hig.update(params_hig)

    X_train_np = prepared.X_train.to_numpy(dtype=np.float32)
    X_test_np = prepared.X_test.to_numpy(dtype=np.float32)

    folds = list(GroupKFold(n_splits=n_splits).split(X_train_np, groups=prepared.groups_train))

    boosters_len = _train_xgb_target(
        X_train_np,
        prepared.y_train_scaled[:, 0],
        folds,
        base_len,
        num_boost_round_len,
        es_rounds_len,
    )
    boosters_hig = _train_xgb_target(
        X_train_np,
        prepared.y_train_scaled[:, 1],
        folds,
        base_hig,
        num_boost_round_hig,
        es_rounds_hig,
    )

    pred_len = _predict_xgb_ensemble(boosters_len, X_test_np)
    pred_hig = _predict_xgb_ensemble(boosters_hig, X_test_np)
    return {
        "boosters_len": boosters_len,
        "boosters_hig": boosters_hig,
        "pred_len_scaled": pred_len,
        "pred_hig_scaled": pred_hig,
    }

xgb_results = train_xgboost_multioutput(prepared_data)
metrics_xgb_windows = evaluate_real_units(
    prepared_data.y_test_real,
    xgb_results["pred_len_scaled"],
    xgb_results["pred_hig_scaled"],
)
print("XGBoost (métricas por ventana en unidades reales):", metrics_xgb_windows)
```

## Catboost

```python
# Entrenamiento CatBoost MultiRMSE compartiendo exactamente los mismos features
from sklearn.model_selection import GroupKFold

try:
    from catboost import CatBoostRegressor, Pool
    _CATBOOST_AVAILABLE = True
except ImportError:
    CatBoostRegressor = None
    Pool = None
    _CATBOOST_AVAILABLE = False

def train_catboost_multioutput(
    prepared: PreparedWindowData,
    params: Optional[Dict] = None,
    n_splits: int = 5,
    es_rounds: int = 800,
    seed: int = SEED,
):
    if not _CATBOOST_AVAILABLE:
        raise ImportError(
            "CatBoost no está instalado. Ejecuta `pip install catboost` para habilitar este modelo."
        )

    base = dict(
        loss_function="MultiRMSE",
        iterations=40000,
        learning_rate=0.02,
        depth=10,
        l2_leaf_reg=2.0,
        border_count=254,
        random_strength=0.5,
        task_type="CPU",
        random_state=seed,
        verbose=False,
    )
    if params:
        base.update(params)

    folds = list(GroupKFold(n_splits=n_splits).split(prepared.X_train, groups=prepared.groups_train))
    models = []
    for tr_idx, va_idx in folds:
        train_pool = Pool(prepared.X_train.iloc[tr_idx], prepared.y_train_scaled[tr_idx])
        valid_pool = Pool(prepared.X_train.iloc[va_idx], prepared.y_train_scaled[va_idx])
        model = CatBoostRegressor(**base)
        model.fit(
            train_pool,
            eval_set=valid_pool,
            use_best_model=True,
            early_stopping_rounds=es_rounds,
            verbose=False,
        )
        models.append(model)

    preds = np.stack([m.predict(prepared.X_test) for m in models], axis=0)
    preds_mean = preds.mean(axis=0)
    return {
        "models": models,
        "pred_len_scaled": preds_mean[:, 0],
        "pred_hig_scaled": preds_mean[:, 1],
    }

if _CATBOOST_AVAILABLE:
    catboost_results = train_catboost_multioutput(prepared_data)
    metrics_catboost_windows = evaluate_real_units(
        prepared_data.y_test_real,
        catboost_results["pred_len_scaled"],
        catboost_results["pred_hig_scaled"],
    )
    print("CatBoost (métricas por ventana en unidades reales):", metrics_catboost_windows)
else:
    catboost_results = None
    print("CatBoost no está disponible; instala el paquete para incluirlo en la comparación.")
```

## **Modelos DL**

```python
import torch, torch.nn as nn, torch.nn.functional as F

class CNN1D_MTL(nn.Module):
    def __init__(self, in_len: int, n_channels=64, p=0.2):
        super().__init__()
        self.fe = nn.Sequential(
            nn.Conv1d(1, n_channels, 7, padding=3), nn.ReLU(),
            nn.Conv1d(n_channels, n_channels, 5, padding=2), nn.ReLU(),
            nn.Conv1d(n_channels, 2*n_channels, 3, padding=1), nn.ReLU(),
            nn.Dropout(p)
        )
        self.proj = nn.Linear(2*n_channels, 128)        # tras GAP
        self.head_len = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))
        self.head_hig = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))

    def forward(self, x_win, len_feat):
        # x_win: (B, W)  -> (B,1,W)
        h = self.fe(x_win.unsqueeze(1))
        h = h.mean(-1)                  # GlobalAvgPool (B, C)
        h = F.relu(self.proj(h))        # (B,128)
        h_cat = torch.cat([h, len_feat.view(-1,1)], dim=1)
        y_len = self.head_len(h_cat).squeeze(1)
        y_hig = self.head_hig(h_cat).squeeze(1)
        return y_len, y_hig
```

```python
class Chomp1d(nn.Module):
    def __init__(self, chomp_size): super().__init__(); self.chomp_size=chomp_size
    def forward(self, x): return x[..., :x.size(-1)-self.chomp_size]

class TCNBlock(nn.Module):
    def __init__(self, C, k=3, d=1, p=0.2):
        super().__init__()
        pad = (k-1)*d
        self.net = nn.Sequential(
            nn.Conv1d(C, C, k, padding=pad, dilation=d), Chomp1d(pad), nn.ReLU(),
            nn.Dropout(p),
            nn.Conv1d(C, C, k, padding=pad, dilation=d), Chomp1d(pad), nn.ReLU()
        )
    def forward(self, x): return x + self.net(x)

class TCN_MTL(nn.Module):
    def __init__(self, in_len, C=64, levels=(1,2,4,8), p=0.2):
        super().__init__()
        self.stem = nn.Conv1d(1, C, 3, padding=1)
        self.blocks = nn.Sequential(*[TCNBlock(C, k=3, d=d, p=p) for d in levels])
        self.proj = nn.Linear(C, 128)
        self.head_len = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))
        self.head_hig = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))
    def forward(self, x_win, len_feat):
        h = self.stem(x_win.unsqueeze(1))
        h = self.blocks(h).mean(-1)          # GAP
        h = F.relu(self.proj(h))
        h_cat = torch.cat([h, len_feat.view(-1,1)], dim=1)
        return self.head_len(h_cat).squeeze(1), self.head_hig(h_cat).squeeze(1)
```

```python
class TinyTransformer_MTL(nn.Module):
    def __init__(self, W, d=128, nhead=4, nlayers=2, patch=8, p=0.1):
        super().__init__()
        self.patch = patch
        self.proj = nn.Conv1d(1, d, kernel_size=patch, stride=patch)   # (B,d,Lp)
        enc_layer = nn.TransformerEncoderLayer(d_model=d, nhead=nhead, dim_feedforward=256,
                                               batch_first=True, dropout=p, activation='gelu')
        self.encoder = nn.TransformerEncoder(enc_layer, num_layers=nlayers)
        self.cls = nn.Parameter(torch.zeros(1,1,d))
        self.hproj = nn.Linear(d, 128)
        self.head_len = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))
        self.head_hig = nn.Sequential(nn.ReLU(), nn.Linear(128+1, 1))

    def forward(self, x_win, len_feat):
        x = self.proj(x_win.unsqueeze(1)).transpose(1,2)  # (B,Lp,d)
        B = x.size(0)
        cls = self.cls.expand(B, -1, -1)
        x = torch.cat([cls, x], dim=1)
        h = self.encoder(x)[:,0]              # token CLS
        h = F.relu(self.hproj(h))
        h = torch.cat([h, len_feat.view(-1,1)], dim=1)
        return self.head_len(h).squeeze(1), self.head_hig(h).squeeze(1)
```
